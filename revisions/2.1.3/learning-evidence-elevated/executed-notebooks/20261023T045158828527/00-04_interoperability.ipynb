# Los cuatro frameworks en un System

Input: value=42 y trail vacío. Proceso: native → LangGraph → OpenAI Agents → Strands, cada uno incrementa y firma el recorrido. Output: value=46 y los cuatro nombres ordenados. Python Runtime ejecuta el control sin inferencia; la campaña también prueba proveedores mixtos. La declaración está en [el ejemplo compartido](../../examples/agentic_systems_studio/src/agentic_systems_studio/lab/interop.py).

```python
import agentic_systems as toolkit
assert toolkit.__version__ == "2.1.3"
```

Este recorrido carga cuatro SDKs. Se declara un plazo de 180 s para incluir su preparación; el plazo habitual de Limits es 10 s y puede agotarse en un kernel frío.

```python
from agentic_systems_studio.lab.interop import build
frameworks = ("native", "langgraph", "openai-agents", "strands")
app = build({name: toolkit.runtime(provider="python-runtime") for name in frameworks})
result = app.run({"value": 42, "trail": []}, limits=toolkit.Limits(timeout_s=180))
assert result.ok and result.data == {"value": 46, "trail": list(frameworks), "validated": True}, result.to_dict()
print(result.data)
print(result.lineage(name="four_frameworks").explain())
```
