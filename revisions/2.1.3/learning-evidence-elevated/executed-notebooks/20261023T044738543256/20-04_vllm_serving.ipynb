```python
import agentic_systems as toolkit

def serving_example():
    artifact = toolkit.model_artifact('fixture/model', base_model='fixture/base', tokenizer='fixture/tokenizer', adapter_path='adapter', revision='fixture-revision', quantization='int8', metadata={'synthetic': True})
    server = toolkit.model_server(artifact, profile='custom', served_model_name='fixture', host='127.0.0.1', port=8123, tool_call_parser='hermes', reasoning_parser='fixture-parser', enable_auto_tool_choice=True, startup_timeout_s=10.0, log_path='fixture-vllm.log', binary='vllm', extra_args=('--dtype', 'auto'), gpu_memory_utilization=0.5, max_model_len=1024, max_num_seqs=2)
    config = toolkit.runtime(provider='vllm-runtime', endpoint=server.endpoint.url, model=server.endpoint.model)
    return {'input': artifact.inspect(), 'process': 'Declare and inspect; no start, health request, model download or inference', 'output': server.inspect(), 'runtime': config.inspect(), 'validation': 'configured-only; lifecycle tested separately with fakes'}
print(serving_example())
```

Después de preparar un entorno GPU, el ciclo es server.start(), server.health(), server.stop(). Este notebook no los llama. La suite offline comprueba ese ciclo con proceso y HTTP simulados. endpoint.url y endpoint.model alimentan runtime; no hay server.runtime ni otra forma de construir el Agent. extra_args no puede duplicar parámetros gestionados.

`model_artifact` describe identidad y procedencia del modelo. `model_server` configura un recurso de serving con ciclo de vida explícito; `start` requiere otro trabajo y acceso al entorno GPU. Este tutorial solo inspecciona la configuración. El Agent recibe el endpoint mediante runtime; no se cambia su gramática.
