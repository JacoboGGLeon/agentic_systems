# Launch the conversational Studio

This notebook starts the Streamlit wrapper for the same system used in `00_conversational_system.ipynb`. It binds only to loopback, waits for health and exposes local and ADA/JupyterLab proxy URLs.


```python
import os

from IPython.display import HTML, display
from agentic_systems_studio import (
    load_studio_environment,
    start_studio_server,
    studio_button_html,
    studio_proxy_url,
)

environment_path = load_studio_environment()
PORT = int(os.getenv('AGENTIC_SYSTEMS_STUDIO_PORT', '8501'))
PROXY_PREFIX = os.getenv(
    'AGENTIC_SYSTEMS_STUDIO_PROXY_PREFIX',
    '/jupyterlab/default/proxy',
)
```

```python
studio_server = start_studio_server(port=PORT)
direct_url = studio_server.local_url
proxy_url = studio_proxy_url(PORT, prefix=PROXY_PREFIX)

print('Health check: ok')
print('Local URL:', direct_url)
print('ADA/JupyterLab proxy:', proxy_url)
print('Log:', studio_server.log_path)
display(HTML(studio_button_html(
    direct_url,
    label='Open Agentic Systems Studio',
    alternate_url=proxy_url,
    alternate_label='Open through ADA/JupyterLab proxy',
)))

if os.getenv('AGENTIC_SYSTEMS_NOTEBOOK_TEST') == '1':
    studio_server.stop()
```

The server intentionally remains active after the launch cell. Stop only the owned process with `studio_server.stop()` when the session ends.
