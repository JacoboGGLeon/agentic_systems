# Composition of complete agentic systems

A CompiledSystem is reused as one computation unit inside a larger System.


```python
from pathlib import Path
import sys
root = Path.cwd()
for candidate in (root / 'src', root / 'examples' / 'agentic_systems_studio' / 'src'):
    if candidate.exists():
        sys.path.insert(0, str(candidate))
```

```python
from agentic_systems_studio import StudioConfig, compose_systems
config = StudioConfig(provider='openai-runtime', framework='agentic-systems')
sequential = compose_systems(('data-quality', 'decision-intelligence'), config, mode='sequential')
parallel = compose_systems(('prompt-security', 'quantitative-analysis'), config, mode='parallel')
print(sequential.compiled.inspect())
print(parallel.compiled.inspect())
```

```python
print(sequential.mermaid())
```

```python
import os
if os.getenv('RUN_LIVE') == '1':
    result = sequential.run('id,name,amount\n1,Ana,10\n2,,20\n2,,20')
    print(result.model_dump(mode='json'))
else:
    print('Composition built and inspected; live provider execution is opt-in.')
```
