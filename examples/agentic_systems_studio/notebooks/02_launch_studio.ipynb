# Launch Agentic Systems Studio

This notebook and the agentic-studio serve CLI use the same public launcher. The server binds only to loopback, waits for the Streamlit health endpoint, writes PID and log files, and exposes the Jupyter proxy URL through an HTML button.


```python
from IPython.display import HTML, display

from agentic_systems_studio import (
    start_studio_server,
    studio_button_html,
    studio_proxy_url,
)

STUDIO_PORT = 8501
PROXY_PREFIX = "/jupyterlab/default/proxy"
```

```python
studio_server = start_studio_server(port=STUDIO_PORT)
proxy_url = studio_proxy_url(STUDIO_PORT, prefix=PROXY_PREFIX)

print("Health check: ok")
print("Local URL:", studio_server.local_url)
print("Jupyter proxy:", proxy_url)
print("PID:", studio_server.process.pid)
print("Log:", studio_server.log_path)
display(HTML(studio_button_html(
    proxy_url,
    label="Abrir Agentic Systems Studio",
)))
```

The server intentionally remains active after the launch cell. Stop it from the same kernel with studio_server.stop(), or interrupt agentic-studio serve when using the CLI foreground mode.
