# Una sola API

Un solo import, una forma de construir cada concepto y RunResult como salida común.

API **2.1.3**. Cada ejemplo muestra una entrada y su resultado. Python Runtime realiza trabajo determinista.

```python
import agentic_systems as toolkit
```

```python
agent = toolkit.Agent("total", toolkit.Action("sum", lambda data: {"total": sum(data["values"])}))
result = agent.run({"values": [2, 3, 5]})
result.data
```

Para generar un proyecto completo, consulta [la CLI](../docs/CLI_213.md). El primer recorrido solo necesita construir, ejecutar y observar el resultado.

Agent y System reciben un flujo: Sequence, Parallel o Graph. Las decisiones son funciones Python en las conexiones. `run(input)` ejecuta; `compile()` es preparación opcional, no otro modo de construir. Skills aportan capacidades. Environment entrega registros y Eval comprueba expectativas. El catálogo completo de API está en `docs/API_213.md`.
