# Fundamentals 00.1 - Runtime API

Objetivo: declarar el runtime de forma explícita con `runtime=...` o `provider="auto"` cuando el entorno tenga una señal clara. Nada de configuración legacy en notebooks nuevos.

````python
import agentic_systems as lab

scheduler = lab.scheduler(
    timeout_s=60,
    max_retries=0,
    max_tool_calls=4,
    max_turns=4,
    max_concurrency=1,
)

auto_runtime = lab.runtime(
    provider="auto",
    scheduler=scheduler,
)

openai_runtime = lab.runtime(
    provider="openai-runtime",
    scheduler=scheduler,
)

bedrock_runtime = lab.runtime(
    provider="bedrock-runtime",
    model=lab.default_model_id(),
    region=lab.default_region(),
    scheduler=scheduler,
)
````
