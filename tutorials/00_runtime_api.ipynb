# Fundamentals 00.1 - Runtime API

Objetivo: declarar el runtime de forma explicita con `runtime=...` o `provider="auto"` cuando el entorno tenga una senal clara. Nada de configuracion legacy en notebooks nuevos.

```python
import agentic_systems as toolkit

scheduler = toolkit.scheduler(
    timeout_s=60,
    max_retries=0,
    max_tool_calls=4,
    max_turns=4,
    max_concurrency=1,
)

auto_runtime = toolkit.runtime(
    provider="auto",
    scheduler=scheduler,
)

openai_runtime = toolkit.runtime(
    provider="openai-runtime",
    scheduler=scheduler,
)

vllm_runtime = toolkit.runtime(
    provider="vllm-runtime",
    scheduler=scheduler,
)

bedrock_runtime = toolkit.runtime(
    provider="bedrock-runtime",
    model=toolkit.default_model_id(),
    region=toolkit.default_region(),
    scheduler=scheduler,
)
```

```python
toolkit.show(auto_runtime.describe(), title="Auto runtime - describe")
```

```python
toolkit.show(openai_runtime.describe(), title="OpenAI runtime - describe")
```

```python
toolkit.show(vllm_runtime.describe(), title="vLLM runtime - describe")
```

```python
toolkit.show(bedrock_runtime.describe(), title="Bedrock runtime - describe")
```

Nota 2.4.1: en notebooks que ejecutan agentes/tools, mostrar primero `toolkit.human_result(...)`; este notebook solo declara runtime y usa `runtime.describe()` como diagnostico.

## Simbolos API explicados

Este notebook se alinea con `docs/API.md` y ensena estos simbolos publicos:

- `toolkit.runtime`: Factory publica para crear RuntimeConfig con provider explicito o auto.
- `toolkit.scheduler`: Factory publica para declarar limites de ejecucion.
- `RuntimeConfig`: Objeto de configuracion del runtime; `describe()` explica la resolucion.
- `SchedulerConfig`: Objeto de limites: timeout, retries, tool calls, turns y concurrencia.
- `default_model_id / default_region`: Defaults explicitos para backends que requieren modelo/region.
- `SUPPORTED_ENGINES / canonical_engine_name / supported_engine_names`: indice canonico de engines validos.
- `BEDROCK_RUNTIME_ENGINE / OPENAI_RUNTIME_ENGINE / PYTHON_DIRECT_ENGINE`: Constantes publicas para evitar strings magicos.
- `__version__`: Version instalada del paquete importado.

- `configure_notebook_environment`: Bootstrap publico para notebooks instalados o ejecutados desde source; se documenta aunque aqui usamos import limpio porque el paquete ya esta instalado.
