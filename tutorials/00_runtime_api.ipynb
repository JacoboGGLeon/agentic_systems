# Fundamentals 00.1 - Runtime API

Objetivo: declarar el runtime de forma explícita con `runtime=...` o `provider="auto"` cuando el entorno tenga una señal clara. Nada de configuración legacy en notebooks nuevos.

```python
import agentic_systems as toolkit

scheduler = toolkit.scheduler(
    timeout_s=60,
    max_retries=0,
    max_tool_calls=4,
    max_turns=4,
    max_concurrency=1,
)

auto_runtime = toolkit.runtime(
    provider="auto",
    scheduler=scheduler,
)

openai_runtime = toolkit.runtime(
    provider="openai-runtime",
    scheduler=scheduler,
)

bedrock_runtime = toolkit.runtime(
    provider="bedrock-runtime",
    model=toolkit.default_model_id(),
    region=toolkit.default_region(),
    scheduler=scheduler,
)
```

```python
toolkit.show(auto_runtime.describe(), title="Auto runtime - describe")
```

```python
toolkit.show(openai_runtime.describe(), title="OpenAI runtime - describe")
```

```python
toolkit.show(bedrock_runtime.describe(), title="Bedrock runtime - describe")
```

Nota 2.4.1: en notebooks que ejecutan agentes/tools, mostrar primero `toolkit.human_result(...)`; este notebook sólo declara runtime y usa `runtime.describe()` como diagnóstico.

## Símbolos API explicados

Este notebook se alinea con `docs/API.md` y enseña estos símbolos públicos:

- `toolkit.runtime`: Factory pública para crear RuntimeConfig con provider explícito o auto.
- `toolkit.scheduler`: Factory pública para declarar límites de ejecución.
- `RuntimeConfig`: Objeto de configuración del runtime; `describe()` explica la resolución.
- `SchedulerConfig`: Objeto de límites: timeout, retries, tool calls, turns y concurrencia.
- `default_model_id / default_region`: Defaults explícitos para backends que requieren modelo/región.
- `SUPPORTED_ENGINES / canonical_engine_name / supported_engine_names`: ?ndice canónico de engines v?lidos.
- `BEDROCK_RUNTIME_ENGINE / OPENAI_RUNTIME_ENGINE / PYTHON_DIRECT_ENGINE`: Constantes públicas para evitar strings mágicos.
- `__version__`: Versión instalada del paquete importado.

- `configure_notebook_environment`: Bootstrap público para notebooks instalados o ejecutados desde source; se documenta aunque aquí usamos import limpio porque el paquete ya está instalado.
