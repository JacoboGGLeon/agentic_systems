# Fundamentals 00.1 - Runtime API

Objetivo: declarar el runtime de forma explícita con `runtime=...` o `provider="auto"` cuando el entorno tenga una señal clara. Nada de configuración legacy en notebooks nuevos.

```python
import agentic_systems as lab

scheduler = lab.scheduler(
    timeout_s=60,
    max_retries=0,
    max_tool_calls=4,
    max_turns=4,
    max_concurrency=1,
)

auto_runtime = lab.runtime(
    provider="auto",
    scheduler=scheduler,
)

openai_runtime = lab.runtime(
    provider="openai-runtime",
    scheduler=scheduler,
)

bedrock_runtime = lab.runtime(
    provider="bedrock-runtime",
    model=lab.default_model_id(),
    region=lab.default_region(),
    scheduler=scheduler,
)
```

```python
lab.show(auto_runtime.describe(), title="Auto runtime - describe")
```

```python
lab.show(openai_runtime.describe(), title="OpenAI runtime - describe")
```

```python
lab.show(bedrock_runtime.describe(), title="Bedrock runtime - describe")
```

Nota 2.4.1: en notebooks que ejecutan agentes/tools, mostrar primero `lab.human_result(...)`; este notebook sólo declara runtime y usa `runtime.describe()` como diagnóstico.
