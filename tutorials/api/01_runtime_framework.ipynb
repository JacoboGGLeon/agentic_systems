# Runtime, scheduler and framework · 2.1.8

A provider selects execution; a framework implements the actor. The scheduler applies limits and retries inside the runtime. All three SDKs are exercised with deterministic Python, without language-model inference.

```python
import agentic_systems as toolkit
```

```python
@toolkit.tool
def echo(value: int) -> dict:
    return {"value": value}
```

```python
def run():
    rows = []
    for framework in ("native", "openai-agents", "strands"):
        worker = toolkit.Agent(
            "operator",
            runtime=toolkit.runtime(provider="python-runtime"),
            framework=toolkit.framework(framework),
            skills=(toolkit.skill(name="echo", tools=(echo,)),),
        )
        result = worker.run({"tool": "echo", "input": {"value": 3}})
        assert result.ok
        rows.append({"framework": framework, "result": result.evidence()})
    return {
        "scope": "real SDKs with deterministic Python; live evidence is separate",
        "rows": rows,
    }
```

```python
report = run()
assert len(report["rows"]) == 3
print([(row["framework"], row["result"]["data"]) for row in report["rows"]])
```
