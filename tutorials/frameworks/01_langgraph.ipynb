# LangGraph: orquestación y pausa · 2.1.4

LangGraph no es un cuarto actor. State/nodes/edges coordinan Agents, Systems y Environments. compile prepara; run puede pausar; resume consume una continuación local al mismo programa.

```python
import pydantic
import agentic_systems as toolkit
assert toolkit.__version__ == '2.1.4'
```

```python
class State(pydantic.BaseModel):
    amount: int | None = None
    approved: bool | None = None
graph = toolkit.Graph(State)
def approval_step(state):
    return {'approved': toolkit.pause({'amount': state.amount, 'question': 'Approve?'})}
graph.add_node('approval', approval_step)
graph.add_edge(toolkit.START, 'approval')
graph.add_edge('approval', toolkit.END)
program = toolkit.Agent('approval', graph=graph).compile()
waiting = program.run({'amount': 280})
assert waiting.status == 'paused'
finished = program.resume(waiting, True)
assert finished.ok and finished.data['approved']
print(finished.data)
```
